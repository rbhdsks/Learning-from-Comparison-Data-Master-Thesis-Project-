Via Grid Search


```python
import numpy as np
coin_tosses=np.array([1,0,1,1,0])
```

```python
def likelihood(p, observations):
    total_probability=1.0
    for result in observations:
        if result==1:
            total_probability=total_probability*p
        else:
            total_probability=total_probability*(1-p)
        
    return total_probability

```

```python
# 0.2,0.4,0.6,0.8
print("likelihood of p when it is 0.2 is", likelihood(0.2,coin_tosses))
print("likelihood of p when its 0.4 is ", likelihood(0.4, coin_tosses))
print("likelihood of p when it is 0.6 is ", likelihood(0.6, coin_tosses))
print("likelihood of p when it is 0.8", likelihood(0.8,coin_tosses))
```

```python
possible_p_values=np.linspace(0,1,1001)
best_p=None
best_likelihood=-1
for p in possible_p_values:
    current_likelihood=likelihood(p,coin_tosses)
    if current_likelihood>best_likelihood:
        best_likelihood=current_likelihood
        best_p=p

print("MLE of P", best_p)
print("Maximum Likelihood", best_likelihood)
```
