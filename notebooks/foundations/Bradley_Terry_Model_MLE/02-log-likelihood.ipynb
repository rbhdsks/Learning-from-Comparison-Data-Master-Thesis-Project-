```python
import numpy as np
from time import perf_counter

coin_tosses=np.array([1,0,1,1,0])

def likelihood(p,observations):
    total_probability=1.0
    for result in observations:
        if result==1:
            total_probability=total_probability*p
        else:
            total_probability=total_probability*(1-p)
    return total_probability
```

```python
def loglikelihood(p,observations):
    total_log_probability=0.0
    for result in observations:
        if result==1:
            total_log_probability=(total_log_probability+ np.log(p))
        else:
            total_log_probability=(total_log_probability+np.log(1-p))
    return total_log_probability
```

```python
p=0.6
ordinary_likelihood=likelihood(p,coin_tosses)
log_of_likelihood=np.log(ordinary_likelihood)
calculated_log_likelihood=loglikelihood(p,coin_tosses)
print("likelihood", ordinary_likelihood)
print("Log of Likelihood", log_of_likelihood)
print("log likelihood function", calculated_log_likelihood)
```

```python
possible_p_values= np.linspace(0.001,0.999,999)
best_p=None
best_log_likelihood=-np.inf
for p in possible_p_values:
    current_log_likelihood=loglikelihood(p,coin_tosses)
    if current_log_likelihood>best_log_likelihood:
        best_log_likelihood=current_log_likelihood
        best_p=p

print("MLE of P:", best_p)
print("Maximum log likelihood",best_log_likelihood)
```

```python
p_from_likelihood= 0.6
p_from_log_likelihood=best_p
p_from_direct_formula=np.mean(coin_tosses)
print("from likelihood:", p_from_likelihood)
print("from log likelihood", p_from_log_likelihood)
print("from direct formula", p_from_direct_formula)
```
