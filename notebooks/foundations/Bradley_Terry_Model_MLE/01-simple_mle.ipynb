```python
import numpy as np 
import matplotlib.pyplot as plt
# where head is 1
coin_tosses=np.array([1,1,0])
print("observations are:", coin_tosses)
print("number of tosses: ", len(coin_tosses))
print("number of head", np.sum(coin_tosses))
```

```python
def likelihood(p):
    probability_of_first_head=p
    probability_of_second_head=p
    probability_of_tail=1-p
    total_probability=(probability_of_first_head*probability_of_second_head*probability_of_tail)
    return total_probability

print("likelihood when probability is 0.2 is ", likelihood(0.2))
print("likelihood when probability is 0.5:", likelihood(0.5))
print("likelihood when probability is 0.67:", likelihood(0.67))
print("Likelihood when probability is 1:", likelihood(1))
```

```python
def likelihood(p):
    return p*p*(1-p)
for p in [0.2,0.5,0.67,0.8,1.0]:
    print("assumed heads probability", p)
    print("assumed tails probability", 1-p)
    print("assumed probability of HHT", likelihood(p))
```

```python
possible_p_values=np.linspace(0,1,1001)
likelihood_values=[]
for p in possible_p_values:
    current_likelihood=likelihood(p)
    likelihood_values.append(current_likelihood)
likelihood_values=np.array(likelihood_values)
best_position=np.argmax(likelihood_values)
p_mle_search=possible_p_values[best_position]
print("MLE Found by searching", p_mle_search)
print("Maximum Likelihood", likelihood_values[best_position])
```

```python
plt.Figure(figsize=(8,5))
plt.plot(
    possible_p_values,likelihood_values,color="royalblue",linewidth=1.5
)
```

```python
number_of_heads=np.sum(coin_tosses)
number_of_tosses=len(coin_tosses)
p_mle=number_of_heads/number_of_tosses
```

```python
plt.Figure(figsize=(8,5))
plt.plot(
    possible_p_values,likelihood_values,color="royalblue",linewidth=1.5
)

plt.axvline(p_mle, color="red",linestyle="--",label=f"mle={p_mle:3f}")
plt.scatter(p_mle,likelihood(p_mle),color="pink", s=70)
plt.xlabel("Possible probability of heads, p")
plt.ylabel("Likelihood L(p)")
plt.title("Likelihood of 3 coin tosses: H,H,T")
plt.grid(alpha=0.25)
plt.legend()
plt.show()
```
