```python
items=np.array(['car','taxi','bus'])
true_scores=np.array([0.8,0.1,-0,9])
```

```python
def sigmoid(value):
    probability=1/(1+np.exp(-value))
    return probability
def bt_probability(item_i,item_j,scores):
    score_difference=scores[item_i]-scores[item_j]
    probability=sigmoid(score_difference)
    return probability
```

```python
p_car_taxi=bt_probability(0,1,true_scores)
```
