```python
import pandas as pd

d = {
    "sales": [
        100000,
        222000,
        1000000,
        522000,
        111111,
        222222,
        1111111,
        20000
    ],
    "city": [
        "Tampa",
        "Tampa",
        "Orlando",
        "Jacksonville",
        "Miami",
        "Jacksonville",
        "Miami",
        "Miami"
    ],
    "size": [
        "Small",
        "Medium",
        "Large",
        "Large",
        "Small",
        "Medium",
        "Large",
        "Small"
    ]
}

```

```python
df = pd.DataFrame(d)
df
```

```python
df['size'].unique()
```

```python
sizes=['Small', 'Medium', 'Large']
```

```python
from sklearn.preprocessing import OrdinalEncoder
```

```python
enc = OrdinalEncoder(categories=[sizes])
```

```python
enc.fit_transform(df[['size']])
```

```python
df.size=enc.fit_transform(df[['size']])
```
