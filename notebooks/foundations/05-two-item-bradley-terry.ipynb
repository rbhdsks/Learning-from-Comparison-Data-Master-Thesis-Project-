```python
import numpy as np
# car beats taxi is 1 and vice versa
observations=np.array([1,0,1,1,0]) 
```

```python
def sigmoid(z):
    probability =1/(1+np.exp(-z))
    return probability
```

```python
def bt_negative_log_likelihood(score_difference,observations):
    p=sigmoid(score_difference)
    nll=-np.sum(observations*np.log(p)+ (1-observations)*(np.log(1-p)))
    return nll
```

```python
print("NLL when d is 1", bt_negative_log_likelihood(-1,observations))
print("NLL when d is 0", bt_negative_log_likelihood(0,observations))
print("NLL when d is -1", bt_negative_log_likelihood(1,observations))
```

```python
def bt_gradient(score_differences, observations):
    predicted_probability=sigmoid(score_differences)
    predicted_car_wins=(len(observations)*predicted_probability)
    observed_car_wins=np.sum(observations)
    gradient=predicted_car_wins-observed_car_wins
    return gradient
```

```python
from time import perf_counter
score_difference=0.0
learning_rate=0.1
number_of_iterations=100
difference_history=[]
nll_history=[]
start_time=perf_counter()
for iterations in range(number_of_iterations):
    gradient=bt_gradient(score_difference, observations)
    score_difference=(score_difference -learning_rate*gradient)
    current_nll=bt_negative_log_likelihood(score_difference, observations)
    difference_history.append(score_difference)
    nll_history.append(current_nll)
end_time =perf_counter()
execution_time =end_time-start_time
estimated_probability=sigmoid(score_difference)
print("estimated score difference", score_difference)
print("estimated car win probability", estimated_probability)
print("final nll", nll_history[-1])
print(f"execution time is {execution_time:.10f} seconds")
```
