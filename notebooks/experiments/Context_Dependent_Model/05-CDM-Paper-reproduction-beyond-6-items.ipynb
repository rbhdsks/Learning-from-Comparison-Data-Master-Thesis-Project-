# 05 — CDM Parameter Recovery: Reproducing Figure 1a and Going Beyond Six Items

## Objective

Seshadri et al. ke Figure 1a mein six-item synthetic data par observations
badhaane se CDM parameter estimation error ghatti hai. Pehle us trend ko
apne implementation se check karna hai. Phir dekhna hai ki item count,
choice-set sampling aur optimizer badalne par recovery kaise badalti hai.

## Existing work se starting point

Mere CDM→CDM notebook mein n=5, choice-set sizes (2, 3), CPU, torch.float64
aur SEED=20260831 use hua hai. Observation scaling mein
m = [1_000, 5_000, 10_000, 50_000] aur 10 runs the.
Item scaling mein n = [5, 10, 20, 40, 80], m=50_000 aur 5 runs the.

Is notebook mein pehle n=6 paper reference hoga; main extension
n = [10, 20, 40] par hogi. n=80 ko baad mein stress test rakhunga.

## Proposed experiments

1. Paper reference: n=6 fix karke m badhao. True CDM matrix, generated
   choices aur fitted matrix se recovery curve banao. Exact paper
   reproduction claim karne se pehle uska data-generation setup match karo.

2. Larger n at fixed data budget: n=[10, 20, 40], m=50_000.
   Same choice-set-size mixture aur fitting protocol rakho.
   Check karo ki parameter RMSE aur test NLL dono kaise badalte hain.

3. m ka effect at larger n: n=20 aur 40 par roughly
   m=[5_000, 20_000, 50_000, 100_000] try karo.
   Pilot runtime dekhkar final upper limit decide karo.
   Har n ke liye true matrix ko m ke across fixed rakho.

4. Sampling schemes: pehle uniform pairs/triples; phir more balanced
   item coverage aur hub-biased sampling. n, m, true matrix aur
   pair/triple proportion same rakho. Pairs-only ko separate
   non-identifiability diagnostic ki tarah dikhao.

5. Optimizers: baseline full-batch Adam; phir GD aur mini-batch method.
   Same dataset aur initialization par final training NLL, gradient/stopping
   status, runtime aur recovery compare karo. Agar fit converge nahi hua,
   us curve ko statistical recovery ka result mat bolo.

## Rough common settings

- Seed family: 20260831 se reproducible, separate parameter/data/split seeds.
- Main choice-set sizes: (2, 3), with their sampling proportion explicitly fixed.
- Baseline fitting: CPU, torch.float64, Adam learning rate 0.05,
  gradient tolerance 1e-5, NLL tolerance 1e-9, patience 50,
  maximum 5_000 epochs — existing CDM settings se start, then verify.
- One pilot run first; 5 runs for feasibility, about 10 independent runs
  for the principal curves if runtime permits.
- m = number of choice observations; a triple choice is one observation.

## Record and plot

For every run log: n, m, true-parameter seed, data seed, sampling scheme,
optimizer, train/test NLL, squared parameter error, off-diagonal RMSE,
epochs/updates, runtime, convergence status and stopping reason.

Plot squared parameter error and RMSE separately. Paper ka 1/m guide
squared-error plot ke liye hai; RMSE par 1/m claim nahi karna.
Mean ± SD ke saath individual run results bhi save karna.

## Questions for the conclusion

- More data se recovery consistently improve hui?
- Fixed m par larger n mein deterioration kitni hui?
- More m dene par larger-n error recover hui, ya optimizer/coverage issue raha?
- Kya test NLL achha tha even when parameter RMSE poor tha?
- Kaunsa result paper Figure 1a ko reproduce karta hai, aur kaunsa
  mera n/sampling/optimizer extension hai?
